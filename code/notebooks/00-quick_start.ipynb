# Quick Start Guide

This notebook contains an example of each default plot provided by the package complex_dynamics.

```python
import complex_dynamics.calculator as dcalc
import complex_dynamics.plot as dplot
import matplotlib.pyplot as plt
import numpy as np
from complex_dynamics.rational_map import RationalMap
```

## Forward Orbit of a Single Point


```python
R = RationalMap.from_latex(r"\frac{2z}{z + 1}")
z0 = np.array([-0.07 + 0.08j])
forward_orbit = dcalc.calculate_forward_orbit(z0, R=R)
fig, ax = dplot.plot_orbit(forward_orbit)
```

## Forward Orbit of Multiple Points (a mesh)

```python
R = RationalMap.from_latex(r"z^2")
Z0 = np.array(
    [0.98 * np.exp(0.025 * 1j), np.exp(0.025 * 1j), 1.02 * np.exp(0.025 * 1j)],
    dtype=np.complex128,
)
forward_orbits = dcalc.calculate_forward_orbit(Z0, R=R, max_iterations=6)
fig, ax = dplot.plot_orbits(forward_orbits)
```

## Divergence Times Static

```python
R = RationalMap.from_latex(r"z^2 - 1")
mesh = dcalc.create_mesh(-2, 2, -2, 2)
divergence_times = dcalc.calculate_divergence_times(mesh, R=R)
fig, ax = dplot.plot_divergence_times(divergence_times)
```

```python
fig, ax = dplot.plot_divergence_times_boundary(divergence_times)
```

## Divergence Times Dynamic

```python
R = RationalMap.from_latex(r"z^2 - 1")
projected_mesh, X, Y, Z = dcalc.create_projected_sphere_mesh(
    theta_pixels=1000, phi_pixels=1000
)
divergence_times = dcalc.calculate_divergence_times(projected_mesh, R=R)
plotter = dplot.dynamic_plot_projected_sphere_divergence_times(
    divergence_times, X, Y, Z
)
```

```python
plotter = dplot.dynamic_plot_projected_sphere_divergence_times_boundary(
    divergence_times, X, Y, Z
)
```

## Convergence Times Static

```python
R = RationalMap.from_latex(r"z^2 - 1")
attractors = np.array([0, np.inf], dtype=np.complex128)
mesh = dcalc.create_mesh(-2, 2, -2, 2)
(convergence_times, convergence_points) = dcalc.calculate_convergence_times(
    mesh, R=R, attractors=attractors
)
fig, ax = dplot.plot_convergence_times(
    convergence_times, convergence_points, attractors=attractors
)
```

```python
fig, ax = dplot.plot_convergence_points_boundary(convergence_points)
```

## Convergence Times Dynamic

```python
R = RationalMap.from_latex(r"z^2 - 1")
R = RationalMap.from_latex(r"z^2 - 1")
attractors = np.array([0, np.inf], dtype=np.complex128)
projected_mesh, X, Y, Z = dcalc.create_projected_sphere_mesh(
    theta_pixels=1000, phi_pixels=1000
)
(convergence_times, convergence_points) = dcalc.calculate_convergence_times(
    projected_mesh, R=R, attractors=attractors
)
plotter = dplot.dynamic_plot_projected_sphere_convergence_times(
    convergence_times, convergence_points, X, Y, Z, attractors=attractors
)
```

```python
plotter = dplot.dynamic_plot_projected_sphere_convergence_points_boundary(
    convergence_points, X, Y, Z
)
```
