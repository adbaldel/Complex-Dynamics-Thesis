# Imports

```python
import complex_dynamics.calculator as dcalc
import complex_dynamics.plot as dplot
import matplotlib.pyplot as plt
import numpy as np
from complex_dynamics.rational_map import RationalMap
```

# Newton's Approximation

## Degree 2

```python
alpha = 1 + 0j
beta = 0 + 1j

P = RationalMap.from_latex(f"(z - {alpha})*(z - {beta})", imag_const="j")

latex = f"z - ({P})/({P.derivative()})"
latex = latex.replace(r"z \left", r"z * \left")

R = RationalMap.from_latex(latex)
print(R)
```

```python
center = 0.0
max_distance_to_center = 10

xmin = center.real - max_distance_to_center
xmax = center.real + max_distance_to_center
ymin = center.imag - max_distance_to_center
ymax = center.imag + max_distance_to_center

extent = [xmin, xmax, ymin, ymax]

xpixels = 400
ypixels = 400

mesh = dcalc.create_mesh(
    xmin, xmax, ymin, ymax, xpixels=xpixels, ypixels=ypixels
)

max_iterations = int(1e2)

attractors = np.array([alpha, beta], dtype=np.complex128)
convergence_times, convergence_points = dcalc.calculate_convergence_times(
    mesh.copy(), R, attractors, max_iterations=max_iterations, tolerance=1e-6
)

# Plot 1: The solid colored basins
fig, ax = dplot.plot_convergence_times(
    convergence_times,
    convergence_points,
    attractors,
    color_peak_position=0.1,
    extent=extent,
)
ax.scatter(
    alpha.real,
    alpha.imag,
    label=rf"$\alpha$",
    marker=".",
    color="black",
    s=50,
    zorder=10,
)
ax.annotate(
    r"$\alpha$",
    (alpha.real, alpha.imag),
    xytext=(10, -10),
    textcoords="offset points",
    ha="center",
    va="center",
    fontsize=8,
    zorder=2,
)
ax.scatter(
    beta.real,
    beta.imag,
    label=rf"$\beta$",
    marker=".",
    color="black",
    s=50,
    zorder=10,
)
ax.annotate(
    r"$\beta$",
    (beta.real, beta.imag),
    xytext=(-10, 10),
    textcoords="offset points",
    ha="center",
    va="center",
    fontsize=8,
    zorder=2,
)
```

```python
projected_sphere_mesh, X, Y, Z = dcalc.create_projected_sphere_mesh()

convergence_times, convergence_points = dcalc.calculate_convergence_times(
    projected_sphere_mesh.copy(),
    R,
    attractors,
    max_iterations=max_iterations,
    tolerance=1e-6,
)

# Plot 1: The solid colored basins
plotter = dplot.dynamic_plot_projected_sphere_convergence_times(
    convergence_times,
    convergence_points,
    X,
    Y,
    Z,
    attractors,
)
```

En general, los puntos que están más cerca de $\alpha$ que de $\beta$ convergen a $\alpha$, y viceversa.
